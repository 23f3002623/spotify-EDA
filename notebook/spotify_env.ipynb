```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("../data/spotify-2023.csv")

# Basic info
print("Shape of dataset:", df.shape)
df.head()
```

```python
# Check missing values
print(df.isnull().sum())

# Drop rows with missing track_name or popularity
df = df.dropna(subset=['track_name', 'popularity'])

# Remove duplicates
df = df.drop_duplicates()

# Show summary
df.info()
```

```python
# Select numeric columns for correlation
num_cols = ['popularity', 'danceability', 'energy', 'speechiness', 'acousticness',
            'instrumentalness', 'liveness', 'valence', 'tempo', 'duration_ms']

corr = df[num_cols].corr()

plt.figure(figsize=(10,6))
sns.heatmap(corr, annot=True, cmap='magma', fmt=".2f")
plt.title("Correlation Heatmap: Features vs Popularity")
plt.show()
```

```python
# Top 10 artists by average popularity
top_artists = df.groupby('artists')['popularity'].mean().sort_values(ascending=False).head(10)
plt.figure(figsize=(10,5))
sns.barplot(x=top_artists.values, y=top_artists.index, palette="cool")
plt.title("Top 10 Artists by Popularity")
plt.xlabel("Average Popularity")
plt.ylabel("Artist")
plt.show()

# Top 10 tracks
top_tracks = df[['track_name', 'artists', 'popularity']].sort_values(by='popularity', ascending=False).head(10)
print(top_tracks)
```

```python
import numpy as np
import random

# Simulate year data (if missing)
df['year'] = np.random.choice(range(2010,2024), size=len(df))

genre_trend = df.groupby(['year','track_genre']).size().reset_index(name='count')
top_genres = genre_trend.groupby('track_genre')['count'].sum().sort_values(ascending=False).head(5).index

plt.figure(figsize=(12,6))
for genre in top_genres:
    subset = genre_trend[genre_trend['track_genre']==genre]
    plt.plot(subset['year'], subset['count'], label=genre)
plt.legend()
plt.title("Genre Growth Over Time")
plt.xlabel("Year")
plt.ylabel("Number of Tracks")
plt.show()
```

```python
plt.figure(figsize=(15,5))

plt.subplot(1,3,1)
sns.scatterplot(data=df, x='valence', y='popularity', alpha=0.5)
plt.title("Valence vs Popularity")

plt.subplot(1,3,2)
sns.scatterplot(data=df, x='tempo', y='popularity', alpha=0.5)
plt.title("Tempo vs Popularity")

plt.subplot(1,3,3)
sns.scatterplot(data=df, x='danceability', y='popularity', alpha=0.5)
plt.title("Danceability vs Popularity")

plt.tight_layout()
plt.show()
```

```python
df['mood'] = df['valence'].apply(lambda x: 'Happy' if x >= 0.5 else 'Sad')

plt.figure(figsize=(6,5))
sns.countplot(data=df, x='mood', palette=['#ff9999','#66b3ff'])
plt.title("Happy vs Sad Songs Distribution")
plt.show()

df.groupby('mood')['popularity'].mean()
```
